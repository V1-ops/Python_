```python
"""1. Flatten Nested Lists ⭐⭐⭐⭐
Problem:-data = [[1, 2], [3, 4, 5], [], [6]]
expected output:-[1, 2, 3, 4, 5, 6]"""

data = [[1, 2], [3, 4, 5], [], [6]]

result = [num for row in data for num in row]

print(result)
# num is at start of the line because the first part of the list comp. is the one which you want to enter into the new list 
# it is read as put the  num to the new list for every row in data and for num in each of the row 
```

```python
"""💡 A useful rule

Whenever you see multiple for clauses in a list comprehension:

[expression
 for A in ...
 for B in ...
]

Mentally rewrite it as:

for A in ...:
    for B in ...:
        result.append(expression)"""
```

```python
"Problem 2 — Remove Duplicates While Preserving Order ⭐⭐⭐⭐"

#solution 

# thinking - we have to remove the duplicates from a list and have to preserve the order of elements
# so we can think of using a data structure which cares about the order and does not allow the duplicates
# and the answer is "Set" for now

nums = [4, 2, 6, 2, 1, 4, 7, 6, 8]
seen = set()
result = []

for num in nums:
    if num not in seen:
        seen.add(num)
        result.append(num)
print(result)
```

```python
"""Problem 3 — Process Chat Messages ⭐⭐⭐⭐

Given

messages = [
    {"role": "user", "content": "Hi"},
    {"role": "assistant", "content": "Hello"},
    {"role": "user", "content": "Tell me about AI"},
    {"role": "assistant", "content": "Sure"},
    {"role": "user", "content": "Thanks"}
]


Return

["Hi", "Tell me about AI", "Thanks"]
"""
messages = [
    {"role": "user", "content": "Hi"},
    {"role": "assistant", "content": "Hello"},
    {"role": "user", "content": "Tell me about AI"},
    {"role": "assistant", "content": "Sure"},
    {"role": "user", "content": "Thanks"}
]

for message in messages:
    if message["role" ]== "user":
        print(message["content"])
```
