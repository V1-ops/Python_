# Module 1 — Tuple Fundamentals

A tuple is an ordered, immutable sequence. Tuples are useful for fixed records, coordinates, database rows, configuration pairs, and multiple return values.

## Syntax
```python
empty = ()
point = (10, 20)
single = (42,)       # comma matters
not_tuple = (42)     # int, parentheses only group
packed = "Ada", 36, "ML"
name, age, job = packed
```

Packing collects comma-separated values; unpacking assigns them to matching targets. A tuple is immutable and a list is mutable. Use a tuple for a fixed structure and a list when items change.

Common mistakes: omitting the comma in a one-item tuple, trying `t[0] = x`, and unpacking the wrong number of values. Interview tip: tuple immutability applies to its slots, not necessarily nested objects.

````python
point = (10, 20)
single = (42,)
print(type(point).__name__, single)

a, b = (1, 2)
print(a, b)
````

## Tuple versus list

| Tuple | List |
|---|---|
| Immutable structure | Mutable structure |
| Good for fixed records | Good for changing collections |
| Can be hashable if all contents are hashable | Unhashable |

### Interview checks
- Why does `(42,)` make a tuple? The comma, not the parentheses.
- Can a tuple contain a list? Yes.
- Can tuple slots be reassigned? No.
