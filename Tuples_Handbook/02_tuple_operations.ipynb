# Module 2 — Tuple Operations

Assume `t = (10, 20, 30, 20)`. Indexing is O(1); negative indexes count from the end. Slicing creates a new tuple in O(k).

```python
t[0]          # 10
t[-1]         # 20
t[1:3]        # (20, 30)
t[::-1]       # reverse
20 in t       # True, O(n)
t + (40,)     # concatenation, new tuple
("x",) * 3   # repetition
```

Built-ins/methods: `len(t)` is O(1); `min`, `max`, and `sum` are O(n); `t.count(x)` and `t.index(x)` are O(n). `index` raises `ValueError` if absent. `min`, `max`, and `sum` require compatible values.

````python
t = (10, 20, 30, 20)
print(t[0], t[-1], t[1:3], t[::-1])
print(len(t), min(t), max(t), sum(t))
print(t.count(20), t.index(30))

for value in t:
    print(value, end=" " )
print()
````

## Practical notes

Concatenation and repetition never mutate the original tuple. A slice also creates a new tuple. For repeated membership checks, use a set if order is irrelevant. Do not confuse `t.index(value)` with dictionary-style lookup: it scans from left to right.
