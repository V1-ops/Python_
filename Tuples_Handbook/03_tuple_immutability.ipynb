# Module 3 — Immutability (Deep Understanding)

Immutability means tuple slots cannot be rebound, added, or removed. It does not mean every referenced object is recursively immutable.

```python
t = (1, 2)
# t[0] = 9             # TypeError
t += (3,)              # creates a new tuple and rebinds t

t = ([1, 2], [3, 4])
t[0].append(5)         # works: the list changed
```

The tuple identity can remain the same while the contents of a nested list change. Identity asks which object (`is`/`id`); equality compares contents (`==`).

## Hashability
A tuple is hashable only when every nested item is hashable. Therefore `(1, 2)` can be a dictionary key, but `([1, 2], 3)` cannot. Mutable objects are unsafe as hash-relevant state.

````python
t = ([1, 2], [3, 4])
old_id = id(t)
t[0].append(5)
print(t, id(t) == old_id)

locations = {(12.9, 77.6): "Bengaluru"}
print(locations[(12.9, 77.6)])
````

### Interview questions
1. Can a tuple contain a list? Yes.
2. Is that tuple deeply immutable? No.
3. Are all tuples dictionary keys? No. Every nested value must be hashable.
4. Why does `t[0].append(5)` work? The list is mutated; the tuple slot is not reassigned.
