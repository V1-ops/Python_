# Module 4 — Packing, Unpacking & Pythonic Tricks

Packing collects values; unpacking assigns them. Multiple assignment and swapping use tuple packing behind the scenes.

```python
record = "Riya", 95, True
student, score, passed = record
a, b = 1, 2
a, b = b, a
head, *middle, tail = (1, 2, 3, 4)
first, _, third = ("keep", "ignore", "keep")
```

A function returning `return min(values), max(values)` returns one tuple. `_` conventionally marks an ignored value. Star unpacking captures a list and must have at most one starred target.

````python
def min_max(values):
    return min(values), max(values)

low, high = min_max([8, 3, 11])
print(low, high)

names = ("Ana", "Bo")
scores = (90, 85)
for index, name in enumerate(names, start=1):
    print(index, name)
for name, score in zip(names, scores, strict=True):
    print(name, score)
````

`enumerate` supplies indexes without manually maintaining a counter. `zip` iterates over parallel sequences and normally stops at the shortest; `strict=True` (Python 3.10+) turns unequal lengths into an error. Unpacking raises `ValueError` when the number of targets does not match.
